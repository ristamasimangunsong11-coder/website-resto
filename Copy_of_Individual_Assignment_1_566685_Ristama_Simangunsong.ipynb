```python
#lucky draw
#three inputs: n, platers, prizes
#pemain is passed as *args
#hadiah is passed as **kwargs
#players and the prizes must be selectd randomly
#each player can win at most once
#each prize can only be selected once
#Use the random module to generate random integers
```

```python
def lucky_draw(n, *players, **prizes):
    import random
    daftar_pemain = list(players)
    daftar_hadiah = list(prizes.keys())

    for i in range(n):
      index_pemain = random.randint(0, len(daftar_pemain)-1)
      index_hadiah = random.randint(0, len(daftar_hadiah)-1)
      pemenang = daftar_pemain[index_pemain]
      hadiah_key = daftar_hadiah[index_hadiah]
      hadiah_value = prizes[hadiah_key]
      print("Winner no.", i+1, "is", pemenang)
      print("Prize: a", hadiah_value, hadiah_key)
```
